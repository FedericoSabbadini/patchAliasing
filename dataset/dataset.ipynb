### Dataset

```python
import pandas as pd
```

```python
# Load historical data
db_url = "Dataset-SolarTechLab.csv"
context_df = pd.read_csv(db_url, delimiter=";")
```

```python
columns = context_df.columns
print(columns)
```

```python
cols = ['Date', 'Power_W', 'Temp_Celsius', 'Irradiance_horizontalPlane', 'Irradiance_inclinedPlane', 'Wind_speed', 'Wind_direction']
context_df.columns = cols
```

```python
#transform Date in a datetime format
context_df["Date"] = pd.to_datetime(
    context_df["Date"],
    format="%d-%b-%Y %H:%M:%S"
)
```

```python
null_val = context_df.isnull().sum().sort_values
print(null_val)
```

```python
# perc null val
perc_null_val = context_df.isnull().sum() / len(context_df) * 100
print(perc_null_val)
```

```python
null_vals = context_df[context_df["Power_W"].isnull()]
print(null_vals)
```

```python
null_number_for_night_hours = null_vals["Date"].dt.hour.value_counts().sort_index()
print(null_number_for_night_hours)
```

```python
# i want to print a row 14:00 case in which power = 0
print(context_df[context_df["Date"].dt.hour == 14])
```

```python
context_df.to_csv("Dataset-SolarTechLabEngineered.csv", index=False)
```
